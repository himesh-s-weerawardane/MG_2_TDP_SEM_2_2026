```python
# ovetime

timestamps = processed_df.date.to_list()
tweets = processed_df.processed.to_list()
```

```python
model = BERTopic(verbose=True)
topics, probs = model.fit_transform(tweets)
topics_over_time = model.topics_over_time(tweets, timestamps)
```

```python
model.visualize_topics_over_time(topics_over_time, topics=[1,2,3,4,5,7,9,12,24,35,44,61,98])
```

```python
from sentence_transformers import SentenceTransformer
from bertopic import BERTopic
from umap import UMAP

# Prepare embeddings
sentence_model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = sentence_model.encode(tweets, show_progress_bar=False)

# Train BERTopic
#topic_model = BERTopic().fit(docs2, embeddings)

topic_model = BERTopic(calculate_probabilities=True, nr_topics = 40)
topics, probs = topic_model.fit_transform(tweets, embeddings)
```

```python
# Reduce dimensionality of embeddings, this step is optional but much faster to perform iteratively:
reduced_embeddings = UMAP(n_neighbors=10, n_components=2, min_dist=0.0, metric='cosine').fit_transform(embeddings)
topic_model.visualize_documents(tweets, reduced_embeddings=reduced_embeddings)
```

```python
import yfinance as yf
import csv
import numpy as np
import pandas as pd
from datetime import datetime
```

```python
# change tickers and date to download the data

yahoo_tickers = [
    "^GSPC", 'AAPL', 'AMT', 'AMZN', 'BA',
       ]

for index, ticker in enumerate(yahoo_tickers):
    with open(csv_path + "/"+ ticker + '.csv', 'w', encoding='UTF8', newline='') as f:
        writer = csv.writer(f)

        data = yf.download(ticker, start="2019-12-31", end="2021-01-01")
        lines = data.to_csv().splitlines()
        reader = csv.reader(lines)

        for row in reader:
            if row[0] == 'Date' :
                writer.writerow(row)
                continue
            writer.writerow(row)
```
